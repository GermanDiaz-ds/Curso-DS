1. Recolección de Datos
2. Exploración y Procesamiento de los datos

#### Histogramas:

```python
# Grafico Variable Age
plt.figure()
#sirve paa crear una nueva figura para el gráfico, lo que permite tener un lienzo separado para dibujar el histograma.
plt.hist(blackfriday['Age'].dropna(), bins=10, color="blue", alpha=0.3)
#sirve para crear un histograma de la columna 'Age' del DataFrame blackfriday.
#dropna() se utiliza para eliminar los valores nulos antes de crear el histograma,
#bins=10 especifica el número de intervalos (barras) en el histograma,
#color="blue" establece el color de las barras del histograma en azul, y 
# alpha=0.3 establece la transparencia de las barras del histograma, donde 0 es completamente transparente
plt.ylabel("Cantidad", size=14)
#sirve para establecer la etiqueta del eje y del gráfico como "Cantidad" con un 
# tamaño de fuente de 14.
plt.xlabel("Edad", size=14)
#sirve para establecer la etiqueta del eje x del gráfico como "Edad" con un 
# tamaño de fuente de 14.
plt.title("Distribucion de la edad de los clientes", size=16, pad=25)
#sirve para establecer el título del gráfico como "Distribucion de la edad de los clientes" con un 
# tamaño de fuente de 16 y un 
# espacio adicional (pad) de 25 unidades entre el título y el gráfico.
plt.gca().spines["top"].set_visible(False)
#sirve para ocultar el borde superior del gráfico, lo que puede mejorar la apariencia visual del gráfico al eliminar elementos innecesarios.
plt.gca().spines["right"].set_visible(False)
#sirve para ocultar el borde derecho del gráfico.
plt.grid(False)
#sirve para desactivar la cuadrícula del gráfico, lo que significa que 
# no se mostrarán líneas de referencia en el fondo del gráfico.
plt.show()
#sirve para mostrar el gráfico generado en la ventana de visualización.
```

Mi pagina favorita para elegir [colores](https://htmlcolorcodes.com/es/tabla-de-colores/)🎨 &nbsp;🖍

#### Distplot:

```python
plt.figure()
sns.displot(blackfriday["Purchase"], color="#5ea88e", kde=True)
#sirve para crear un histograma utilizando la función displot() de Seaborn.
#kde=True agrega una estimación de densidad de kernel (KDE) al histograma, 
# lo que proporciona una representación suave de la distribución de los datos.
plt.xlabel('Monto de la compra')
#sirve para establecer la etiqueta del eje x del gráfico como "Monto de la compra".
plt.ylabel('Frequency')
#sirve para establecer la etiqueta del eje y del gráfico como "Frequency".
sns.despine()
#sirve para eliminar los bordes superior e izquierdo del gráfico.
plt.show()
#sirve para mostrar el gráfico generado en la ventana de visualización.
```

#### Boxplot:

```python
boxplot_blackfriday = blackfriday.melt(id_vars='City_Category', value_vars=['Purchase'])
#sirve para transformar el DataFrame blackfriday de 
# formato ancho a formato largo utilizando la función melt() de pandas.
#id_vars='City_Category' especifica que la columna 'City_Category' se mantendrá como una variable identificadora, 
# mientras que value_vars=['Purchase'] indica que la columna 'Purchase' se convertirá en una variable de valor. 
```

```python
boxplot_blackfriday.head()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x="variable", y="value", data=boxplot_blackfriday, palette="Set2", hue='City_Category')
#sirve para crear un diagrama de caja (boxplot) utilizando la función boxplot() de Seaborn.
#x="variable" especifica que la columna "variable" se utilizará para el eje x del gráfico,
#y="value" indica que la columna "value" se utilizará para el eje y del gráfico,
#data=boxplot_blackfriday especifica el DataFrame que contiene los datos
#palette="Set2" establece la paleta de colores utilizada para los diferentes grupos en el gráfico, y 
#hue='City_Category' indica que se utilizará la columna "City_Category" para diferenciar los grupos en el gráfico, 
# asignando colores distintos a cada categoría de ciudad       
plt.title('Monto segun categoria de ciudad', size=14, pad=15)
plt.legend(loc='best', frameon=False)
#loc 'best' coloca la leyenda en la mejor posición disponible
#frameon=False elimina el marco alrededor de la leyenda
plt.xlabel('Ciudad', size=12)
plt.ylabel('Monto', size=12)
sns.despine()
#despine() elimina los bordes superior e izquierdo del gráfico
plt.show()
```

#### Pairplot

```python
blackfriday.columns
```

```python
# Revisar la relación entre Stay_In_Current_City_Years, Age, Purchase y Gender
pairplot_data = blackfriday[['Stay_In_Current_City_Years', 'Age', 'Purchase', 'Gender']].sample(
    n=min(5000, len(blackfriday)), random_state=42
)
sns.pairplot(pairplot_data, hue='Gender', palette='PuRd', corner=True)

#se puede observar que la variable "Purchase" tiene una distribución sesgada hacia la derecha,
# lo que indica que la mayoría de las compras se concentran en valores más bajos, 
# mientras que hay algunas compras excepcionales con montos significativamente más altos.
```

```python
corr = blackfriday[['Age', 'Stay_In_Current_City_Years', 'Purchase']].corr(method = 'pearson')
```

```python
sns.heatmap(corr, cmap='YlGnBu', annot=True);
#con este grafico se puede observar la correlación entre las variables Age, Stay_In_Current_City_Years y Purchase.
#se puede observar que la correlación entre Age y Purchase es negativa, 
# lo que indica que a medida que aumenta la edad, el monto de la compra tiende a disminuir.
```
