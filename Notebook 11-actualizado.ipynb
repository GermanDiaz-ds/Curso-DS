#### 1) TRANSFORMACION DE VARIABLES

- La mayoría de los algoritmos de machine learning no admiten `strings` como variables y requieren que las variables sean numericas. Por ese motivo, es necesario convertir las variables categoricas en su representación numerica. Para esto hay varias opciones que iremos viendo.

**`LabelEncoder()`**

```python
from sklearn.preprocessing import LabelEncoder 
# Con Label encoder lo que hacemos es cambiar una categoria que tenia nombre 
# por ejemplo de ciudades en que las ponga 0,1,2,3, ect
```

```python
test_encoder = LabelEncoder() #Ejecutamos la función
```

```python
blackfriday.columns
```

```python

blackfriday['City_Category'] = test_encoder.fit_transform(blackfriday['City_Category'])
```

```python
blackfriday.head(10)
```

**`get_dummies()`**

```python
pd.get_dummies(blackfriday["Gender"])
# es necesario dropear una columna para evitar la multicolinealidad, ya que si tenemos 2 columnas que son iguales, una de ellas es redundante y no aporta información al modelo. Por lo tanto, se recomienda dropear una columna para evitar la multicolinealidad. En este caso, se dropea la columna "Gender
```

```python
blackfriday[['female', 'male']] = pd.get_dummies(blackfriday["Gender"])
```

```python
blackfriday.head()
```

**`OneHotEncoder()`**

```python
#onehotencoder sirve para convertir variables categóricas en variables dummy (0 o 1)
from sklearn.preprocessing import OneHotEncoder
```

```python
blackfriday['Gender'].unique()
```

```python
gender_encoder = OneHotEncoder()
```

```python
gender_encoder.fit_transform(blackfriday[['Gender']]).toarray()
```

```python
gender_encoder.categories_
```

```python
niveles = gender_encoder.categories_[0].tolist() #get_features_names_out usar ese
print(niveles)
```

```python
one_hot_gender = pd.DataFrame(gender_encoder.fit_transform(blackfriday[['Gender']]).toarray(), columns=niveles)
```

```python
one_hot_gender
```

```python
new_df = pd.concat([blackfriday, one_hot_gender], axis=1) #puedo usar join
```

```python
new_df.head()
#quedo pegado el booleano del getdummy y la columna original, por lo que es necesario dropear la columna original para evitar la multicolinealidad. En este caso, se dropea la columna
```

**`binning`**


```python
#binning sirve para agrupar valores continuos en intervalos o categorías
print(blackfriday.Age.min())
print(blackfriday.Age.max())
```

```python
blackfriday["Age"] = pd.to_numeric(blackfriday["Age"], errors='coerce')
```

```python
bin_age = [10, 17, 70, 80] #parentesis incluye el 10 y corchete lo excluye, por lo que el 17 queda en el primer grupo y el 70 en el segundo grupo. El 80 queda en el tercer grupo.
labels = ["Adolescente", "Adulto", "Anciano"]

age_categories = pd.cut(blackfriday["Age"], bins=bin_age, labels=labels)
```
